# Cobertura de Código

Cobertura ≠ calidad.

````python
def clasificar(x):
    if x > 0:
        return 'positivo'
    else:
        return 'no positivo'
````

````python
def test_clasificar():
    assert clasificar(5) == 'positivo'
````

## Pregunta
¿Qué falta probar?

````python
def test_negativo():
    assert clasificar(-1) == 'no positivo'
````

## Reflexión: ¿100% coverage garantiza corrección?

**Respuesta: NO.** El 100% de cobertura no es sinónimo de corrección.

### Demostración práctica:

```python
# Función con bugs pero 100% coverage
def calcular_descuento(precio, es_vip):
    if es_vip:
        return precio * 0.9  # 10% descuento
    else:
        return precio * 0.95  # 5% descuento (BUG: debería ser sin descuento)

# Tests que logran 100% coverage pero no detectan el bug
def test_vip():
    assert calcular_descuento(100, True) == 90

def test_no_vip():
    assert calcular_descuento(100, False) == 95  # Test incorrecto, debería fallar
```

### Distinción clave:

**Cobertura:** Mide qué porcentaje del código es ejecutado por los tests
- Líneas ejecutadas
- Ramas cubiertas  
- Funciones invocadas

**Calidad:** Mide qué tan bien los tests validan el comportamiento correcto
- Casos límite probados
- Valores correctos verificados
- Errores detectados
- Comportamiento real validado

### Ejemplos de alta cobertura, baja calidad:

1. **Tests que solo verifican tipos:**
```python
def test_devuelve_int():
    result = suma(2, 3)
    assert isinstance(result, int)  # 100% coverage, pero no verifica valor
```

2. **Tests que solo verifican que no hay excepciones:**
```python
def test_no_explota():
    procesar_datos(datos_complejos)  # 100% coverage, pero no verifica resultado
```

3. **Tests con valores incorrectos:**
```python
def test_calculo():
    assert area_circulo(1) == 3.14  # 100% coverage, pero test incorrecto
```

### ¿Por qué 100% coverage ≠ corrección?

1. **Tests pueden ser incorrectos:** Los tests mismos pueden tener bugs
2. **Cobertura no mide aserciones:** Solo mide ejecución, no validación
3. **Casos límite pueden faltar:** 100% coverage sin probar edge cases
4. **Lógica de negocio puede estar mal:** Tests pueden pasar con implementación incorrecta

### Mejor enfoque:

1. **Cobertura como herramienta, no objetivo:** Úsala para encontrar código no probado
2. **Calidad sobre cantidad:** Mejor 70% coverage con tests robustos que 100% con tests débiles
3. **Revisión crítica de tests:** ¿Realmente validan el comportamiento correcto?
4. **Tests basados en especificaciones:** Escribir tests que reflejen los requisitos reales

### Conclusión:

La cobertura es una métrica útil pero engañosa. **100% coverage solo dice "ejecutamos todo el código", no "el código es correcto"**. La calidad de las pruebas se mide por su capacidad para detectar errores reales, no por el porcentaje de líneas cubiertas.
